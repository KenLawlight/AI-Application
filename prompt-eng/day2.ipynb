````python
import os
from dotenv import load_dotenv
from openai import OpenAI
````

````python
load_dotenv(override=True)
openai_api_key = os.getenv('OPENAI_API_KEY')

client = OpenAI(api_key=openai_api_key)
````

````python
def get_response(prompt):
    response =client.chat.completions.create(model='gpt-4.1-mini', messages=[{"role": "user", "content": prompt}],max_tokens=150, temperature=0)
    return response.choices[0].message.content
````

````python
story = "The Canopy of Tomorrow: A StoryFor centuries, the lush, emerald canopy of the Amazon basin stood as the world’s most vibrant living library. Deep within this vibrant sanctuary, a diverse chorus of howler monkeys, iridescent macaws, and countless uncatalogued plants thrived in a carefully balanced ecosystem. For the indigenous communities who called these woods home, the forest wasn't just a collection of trees; it was their pharmacy, their pantry, and the heartbeat of their culture."

prompt = f"""Complete the story delimited by triple backticks. 
 ```{story}```"""


response = get_response(prompt)

print("\n Original story: \n", story)
print("\n Generated story: \n", response)
````
