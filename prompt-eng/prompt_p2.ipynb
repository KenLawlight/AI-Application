```python
import os
from dotenv import load_dotenv
from openai import OpenAI
```

```python
load_dotenv(override=True)
openai_api_key = os.getenv('OPENAI_API_KEY')

client = OpenAI(api_key=openai_api_key)
```

```python
def get_response(prompt):
    response =client.chat.completions.create(model='gpt-4.1-mini', messages=[{"role": "user", "content": prompt}],max_tokens=150, temperature=0)
    return response.choices[0].message.content
```

```python
prompt = """
     Q: Extract the odd numbers from {1, 3, 7, 12, 19}. A: Odd numbers = {1, 3, 7, 19}
     Q: Extract the odd numbers from {3, 5, 11, 12, 16}. A:
"""


response = get_response(prompt)
print(response)
```
